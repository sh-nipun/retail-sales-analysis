```python
# Convert Order Date and Ship Date columns from string to datetime format
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%m/%d/%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%m/%d/%Y')

# Verify the conversion worked correctly
print(df['Order Date'].dtype)
df['Order Date'].head()
```

```python
# Standardize column names: remove spaces and convert to lowercase with underscores
# This makes column names easier to use in SQL queries later (e.g. Order_Date instead of "Order Date")
df.columns = df.columns.str.strip().str.replace(' ', '_')

# Verify the new column names
print(df.columns.tolist())
```

```python
import os

# Create the data/processed directory if it doesn't already exist
os.makedirs('../data/processed', exist_ok=True)

print("Directory created (or already exists)!")
```

```python
# Save the cleaned dataset to a new CSV file for later use
df.to_csv('../data/processed/cleaned_superstore.csv', index=False)

print("Cleaned data saved successfully!")
```

```python
import sqlite3

# Create (or connect to) a SQLite database file inside data/processed folder
conn = sqlite3.connect('../data/processed/superstore.db')

# Write the cleaned DataFrame into a table named 'sales' inside the database
# if_exists='replace' ensures re-running this cell won't create duplicate tables
df.to_sql('sales', conn, if_exists='replace', index=False)

print("Data successfully written to SQLite database!")
```

```python
# Query: Calculate total sales grouped by month
monthly_trend_query = """
SELECT strftime('%Y-%m', Order_Date) AS month, 
       SUM(Sales) AS total_sales,
       SUM(Profit) AS total_profit
FROM sales
GROUP BY month
ORDER BY month
"""

# Run the query and store result in a DataFrame
monthly_trend = pd.read_sql(monthly_trend_query, conn)
monthly_trend.head(10)
```

```python
# Query: Find the top 10 products by total sales amount
top_products_query = """
SELECT Product_Name, 
       SUM(Sales) AS total_sales,
       SUM(Quantity) AS total_quantity
FROM sales
GROUP BY Product_Name
ORDER BY total_sales DESC
LIMIT 10
"""

top_products = pd.read_sql(top_products_query, conn)
top_products
```

```python
# Query: Compare total sales and profit across regions
region_performance_query = """
SELECT Region, 
       SUM(Sales) AS total_sales, 
       SUM(Profit) AS total_profit,
       ROUND(SUM(Profit) * 100.0 / SUM(Sales), 2) AS profit_margin_pct
FROM sales
GROUP BY Region
ORDER BY total_profit DESC
"""

region_performance = pd.read_sql(region_performance_query, conn)
region_performance
```

```python
# Query: Analyze profit margin by product category
category_breakdown_query = """
SELECT Category, 
       SUM(Sales) AS total_sales, 
       SUM(Profit) AS total_profit,
       ROUND(SUM(Profit) * 100.0 / SUM(Sales), 2) AS profit_margin_pct
FROM sales
GROUP BY Category
ORDER BY total_profit DESC
"""

category_breakdown = pd.read_sql(category_breakdown_query, conn)
category_breakdown
```
