```python
# Import pandas library for data manipulation
import pandas as pd

# Quick check to confirm environment setup is working
print("Setup working!")
```

```python
# Load the raw Superstore dataset using its full file path
df = pd.read_csv(r"C:\Users\Kirito\Documents\Data analysis project\retail-sales-analysis\Sample - Superstore.csv", encoding='latin1')

# Display first 5 rows to get an overview of the data structure
df.head()
```

```python
# Check current data type of Order Date column
print(df['Order Date'].dtype)
print(df['Order Date'].head())
```

```python
# Convert Order Date and Ship Date columns from string to datetime format
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%m/%d/%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%m/%d/%Y')

# Verify the conversion worked correctly
print(df['Order Date'].dtype)
df['Order Date'].head()
```

```python
# Standardize column names: remove spaces and convert to lowercase with underscores
# This makes column names easier to use in SQL queries later (e.g. Order_Date instead of "Order Date")
df.columns = df.columns.str.strip().str.replace(' ', '_')

# Verify the new column names
print(df.columns.tolist())
```

```python
import os

# Create the data/processed directory if it doesn't already exist
os.makedirs('../data/processed', exist_ok=True)

print("Directory created (or already exists)!")
```

```python
# Save the cleaned dataset to a new CSV file for later use
df.to_csv('../data/processed/cleaned_superstore.csv', index=False)

print("Cleaned data saved successfully!")
```

```python
import sqlite3

# Create (or connect to) a SQLite database file inside data/processed folder
conn = sqlite3.connect('../data/processed/superstore.db')

# Write the cleaned DataFrame into a table named 'sales' inside the database
# if_exists='replace' ensures re-running this cell won't create duplicate tables
df.to_sql('sales', conn, if_exists='replace', index=False)

print("Data successfully written to SQLite database!")
```

```python
# Query: Calculate total sales grouped by month
monthly_trend_query = """
SELECT strftime('%Y-%m', Order_Date) AS month, 
       SUM(Sales) AS total_sales,
       SUM(Profit) AS total_profit
FROM sales
GROUP BY month
ORDER BY month
"""

# Run the query and store result in a DataFrame
monthly_trend = pd.read_sql(monthly_trend_query, conn)
monthly_trend.head(10)
```

```python
# Query: Find the top 10 products by total sales amount
top_products_query = """
SELECT Product_Name, 
       SUM(Sales) AS total_sales,
       SUM(Quantity) AS total_quantity
FROM sales
GROUP BY Product_Name
ORDER BY total_sales DESC
LIMIT 10
"""

top_products = pd.read_sql(top_products_query, conn)
top_products
```

```python
# Query: Compare total sales and profit across regions
region_performance_query = """
SELECT Region, 
       SUM(Sales) AS total_sales, 
       SUM(Profit) AS total_profit,
       ROUND(SUM(Profit) * 100.0 / SUM(Sales), 2) AS profit_margin_pct
FROM sales
GROUP BY Region
ORDER BY total_profit DESC
"""

region_performance = pd.read_sql(region_performance_query, conn)
region_performance
```

```python
# Query: Analyze profit margin by product category
category_breakdown_query = """
SELECT Category, 
       SUM(Sales) AS total_sales, 
       SUM(Profit) AS total_profit,
       ROUND(SUM(Profit) * 100.0 / SUM(Sales), 2) AS profit_margin_pct
FROM sales
GROUP BY Category
ORDER BY total_profit DESC
"""

category_breakdown = pd.read_sql(category_breakdown_query, conn)
category_breakdown
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set a clean visual style for all charts
sns.set_style("whitegrid")

# Create a line chart showing monthly sales trend
plt.figure(figsize=(12, 5))
sns.lineplot(data=monthly_trend, x='month', y='total_sales', marker='o')
plt.xticks(rotation=45)
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.tight_layout()

# Save the chart as an image for the README
plt.savefig('../visuals/monthly_sales_trend.png', dpi=150)
plt.show()
```

```python
# Create a horizontal bar chart for top 10 products by sales
plt.figure(figsize=(10, 6))
sns.barplot(data=top_products, x='total_sales', y='Product_Name', palette='viridis')
plt.title('Top 10 Products by Sales')
plt.xlabel('Total Sales')
plt.ylabel('Product Name')
plt.tight_layout()

plt.savefig('../visuals/top_products.png', dpi=150)
plt.show()
```

```python
# Reshape data for grouped bar chart (sales and profit side by side)
region_melted = region_performance.melt(id_vars='Region', value_vars=['total_sales', 'total_profit'],
                                          var_name='Metric', value_name='Amount')

plt.figure(figsize=(10, 6))
sns.barplot(data=region_melted, x='Region', y='Amount', hue='Metric')
plt.title('Region-wise Sales vs Profit')
plt.xlabel('Region')
plt.ylabel('Amount')
plt.tight_layout()

plt.savefig('../visuals/region_sales_vs_profit.png', dpi=150)
plt.show()
```

```python
# Bar chart showing profit margin percentage by category
plt.figure(figsize=(8, 5))
sns.barplot(data=category_breakdown, x='Category', y='profit_margin_pct', palette='coolwarm')
plt.title('Profit Margin by Category (%)')
plt.xlabel('Category')
plt.ylabel('Profit Margin (%)')
plt.tight_layout()

plt.savefig('../visuals/category_profit_margin.png', dpi=150)
plt.show()
```
